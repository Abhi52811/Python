```python
# .read() — Read entire content

f = open('../clutteredFolder/myfile.txt', 'r')
print(f)
text = f.read()
print(text)
f.close()
```

```python
# .readline() — Read one line at a time

f = open('../clutteredFolder/myfile.txt', "r")
line1 = f.readline()
line2 = f.readline()
f.close()
print(line1)
print(line2)
```

```python
# .readlines() — Returns list of all lines
f = open('../clutteredFolder/myfile.txt', "r")
lines = f.readlines()
print(lines)
f.close()
```

```python
# .write()

f = open('../clutteredFolder/myfile.txt', 'a')
f.write('\nAbhishek here')
f.close()
```

```python
# .writelines() — Write multiple lines

lines = ["Line1\n", "Line2\n"]
f = open('../clutteredFolder/myfile.txt', "w")
f.writelines(lines)
f.close()
```

```python
# Appending to a File
f = open('../clutteredFolder/myfile.txt', "a")
f.write("\nThis is new data.")
f.close()
```

with -> context manager

```python
# Using with (Auto-close)
# Recommended way to handle files

# when using "with" we don't need to close the file specifically
# it will automatically close it
with open('../clutteredFolder/myfile.txt', 'a') as f:
  f.write("\nHey I am inside with")
```

```python
with open('../clutteredFolder/myfile.txt', "r") as src, open('../clutteredFolder/myfile5.txt', "w") as dest:
    dest.write(src.read())
```

```python
# with -> context manager --> handles closing of your file
full_file_path = "../clutteredFolder/myfile1.txt"
with open(full_file_path) as f:
    lines = f.readlines()
    print(lines)
```

```python
# with -> context manager --> handles closing of your file
with open(full_file_path, "w") as f:
    f.write("hello, how are you.\n")# with -> context manager --> handles closing of your file

with open(full_file_path) as f:
    lines = f.read()
    print(lines)
```

```python
a = ["hello, how are you.\n", "hey,      ", "hello"]
with open(full_file_path, "w") as f:
    f.writelines(a)
    
with open(full_file_path) as f:
    lines = f.read()
    print(lines)
```

```python
with open(full_file_path, "a") as f:
    f.write("new statement added.\n")

with open(full_file_path) as f:
    lines = f.read()
    print(lines)
```

```python
# Read+write

with open(full_file_path, "r+") as f:
    f.write("new statement added.\n\n")

with open(full_file_path) as f:
    lines = f.read()
    print(lines)
```
