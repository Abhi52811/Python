# Local

- Variables defined inside a function.
- They exist only within the function.
- Changes affect only local copy.

```python
x = 10 
def my_function():
  x=5
  print(f"The local x is {x}")
print(f"The global x is {x}")
my_function()
print(f"The global x is {x}")
```

# Global

- Defined outside any function.
- Accessible inside functions (read-only by default).
- To modify global variable inside a function, use `global` keyword.

```python
x = 10  # global variable
y = 20

def my_function():
  global x
  x = 5  # this will change the value of the global variable x
  y = 5  # local variable
  print(f"The vale x is {x}")
  print(f"The value y is {y}")

print(f"The global x is {x}")
print(f"The global y is {y}")
my_function()
print(f"The global x is {x}")
print(f"The global y is {y}")
```

# Non local

- Used inside nested functions.
- Refers to variables in the nearest enclosing (non-global) scope.
- Modify outer function’s variable.

```python
def outer():
    x = 5
    def inner():
        nonlocal x
        x = 10  # changes x in outer
    inner()
    print(x)  # prints 10

outer()
```
