## Variable Scope in Functions

Local Variables

```python
def func():
    a = 10  # local variable
    print(a)

func()
print(a)  # Error! 'a' is not defined outside
```

Global Variables

```python
x = 5  # global variable

def func():
    global x
    x = 10  # modifies global x
    print(x)

func()  # prints 10
print(x)  # prints 10, global changed
```

Nonlocal Variables

```python
def outer():
    x = 5
    def inner():
        nonlocal x
        x = 10  # changes x in outer
    inner()
    print(x)  # prints 10

outer()
```

# Passing Arguments: By Value or Reference

Python passes arguments by object reference (sometimes called “pass-by-assignment”):
- For mutable objects (like lists, dicts), changes inside function affect original.
- For immutable objects (like int, str, tuple), reassignment inside function doesn’t affect original.

```python
# Example with mutable:
def append_item(lst):
    lst.append(4)

my_list = [1, 2, 3]
append_item(my_list)
print(my_list)  # [1, 2, 3, 4] changed!
```

```python
# Example with immutable:

def change_num(n):
    n = 10

x = 5
change_num(x)
print(x)  # 5 unchanged
```

# Special Argument Syntax: *args and **kwargs

*args — Variable Positional Arguments
- Collects extra positional arguments into a tuple.
- Useful when number of positional args unknown.

```python
def func(*args):
    for arg in args:
        print(arg)

func(1, 2, 3)
```

**kwargs — Variable Keyword Arguments
- Collects extra keyword arguments into a dictionary.
- Useful for optional named parameters.

```python
def func(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")

func(a=1, b=2)
```

Combined Use

```python
def func(a, b, *args, **kwargs):
    print(a, b)
    print(args)
    print(kwargs)

func(1, 2, 3, 4, x=5, y=6)
```

# Positional-Only and Keyword-Only Arguments

- Positional-only arguments: indicated by / in function signature, must be passed by position. Any parameter before / must be passed positionally. Parameters after / can be passed either positionally or as keywords.
- Keyword-only arguments: indicated by * in signature, must be passed by keyword. Any parameter after * must be passed as a keyword (c=value). Parameters before * can be passed either positionally or as keywords

```python
def f(a, b, /, c, *, d):
    print(a, b, c, d)

f(1, 2, 3, d=4)  # valid
f(1, 2, c=3, d)  # invalid, a and b positional-only
```

Mimic function Overloading

```python
from functools import singledispatch

@singledispatch
def fun(arg):
    print("Default implementation:", arg)

@fun.register
def _(arg: int):
    print("Integer implementation:", arg)

@fun.register
def _(arg: str):
    print("String implementation:", arg)

fun(10)     # Integer implementation: 10
fun("hi")   # String implementation: hi
fun(3.14)   # Default implementation: 3.14
```

#  Advanced Ways to Pass Values

 Passing functions as arguments

```python
def apply_func(f, x):
    return f(x)

print(apply_func(lambda x: x * 2, 5))  # 10
```

 Passing unpacked iterables as arguments

```python
def add(a, b, c):
    print(a + b + c)

nums = [1, 2, 3]
add(*nums)  # unpack list to positional arguments

kwargs = {'a':1, 'b':2, 'c':3}
add(**kwargs)  # unpack dict to keyword arguments
```

# Getters and setters Introduction

```python
class myClass:
    def __init__(self,value):
        self._value= value
    def show(self):
        print(f"Value is {self._value}")
    @property
    def value(self):
        return self._value
```

```python
obj = myClass(10)
obj.show()
```

```python
class MyClass:
  def __init__(self, value):
      self._value = value
    
  def show(self):
    print(f"Value is {self._value}")
    
  # using @property it will become a getter 
  @property
  def ten_value(self):
      return 10* self._value
    
  # using @ten_value.setter it will become a setter 
  @ten_value.setter
  def ten_value(self, new_value):
      self._value = new_value/10
```

```python
obj = MyClass(10)
obj.ten_value = 67
print(obj.ten_value)
obj.show()
```
