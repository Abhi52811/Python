```python
def temp():
    print("Invalid")

# anonymous lambda function
# lambda: print("Invalid")
print((lambda: print("Invalid"))())
```

```python
print((lambda x, y: x + y)(3,5))
```

```python
def add(x, y):
    return x + y

# Lambda takes a function and arguments, calls the function
print(type((lambda func: func(10, 5))(add)))  # ➜ 15
print((lambda func: func(10, 5))(add))  # ➜ 15
```

```python
def double(x):
    return x*2

print(double(5))
```

```python
double = lambda x : x*2
print(double(5))
print(type(double))
```

```python
cube = lambda x : x*x*x
print(cube(5))
print(type(cube))
```

```python
avg = lambda x,y : (x+y)/2
print(avg(5,6))
print(type(avg))
```

```python
# passing the function as parameter
#       (cube, 5 )
def appl(fx,value):
    return 6 + fx(value)
#          6 + cube(5)

print(appl(cube,5))
```

```python
def appl(fx,value):
    return 6 + fx(value)

print(appl(lambda x: x*x*x,5))
```
