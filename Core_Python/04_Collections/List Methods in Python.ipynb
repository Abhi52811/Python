```python
l=[2,1,9,7,3,4,6]
print(l)
```

```python
# append(x)
# Adds a single item to the end of the list.

lst = [1, 2]
lst.append(3)
print(lst)  # [1, 2, 3]

lst1 = [1, 2]
lst2 = [3,4]
lst1.append(lst2)
print(lst1)
```

```python
# extend(iterable)
# Adds each element of an iterable to the end of the list.

lst = [1, 2]
lst.extend([3, 4])
print(lst)  # [1, 2, 3, 4]

lst1 = [1, 2]
lst2 = [3,4]
lst1.extend(lst2)
print(lst1)
```

```python
l1=[1,2,3]
l2=[4,5,6]
l1 = l1+l2
print(l1)
```

```python
# insert(i, x)
# Inserts x at position i.

lst = [1, 3]
lst.insert(1, 2)
print(lst)  # [1, 2, 3]
```

```python
# reverse()
l=[2,1,9,7,3,4,6]
l.reverse()
print(l)
```

```python
# sort()
# change in the original list
# Sorts the list in place.

l=[2,1,9,7,3,4,6]
l.sort()
print(l)
```

```python
# sort() in reverse
# change in the original list
l=[2,1,9,7,3,4,6]
l.sort(reverse=True)
print(l)
```

```python
l=[2,1,9,7,3,4,6]
l.sort()
print(l)
print("Index of 7 : ",l.index(7))
print("Count of 7 : ",l.count(7))
l.sort(reverse=True)
print(l)
print("Index of 7 : ",l.index(7))
print("Count of 7 : ",l.count(7))
```

```python
# remove(x)
# Removes the first occurrence of x.

lst = [1, 2, 3, 2]
lst.remove(2)
print(lst)  # [1, 3, 2]
```

```python
# pop([i])
# Removes and returns element at index i. If no index is specified, removes the last element.

lst = [1, 2, 3]
x = lst.pop()
print(x)    # 3
print(lst)  # [1, 2]
```

```python
# clear()
# Removes all items from the list.

lst = [1, 2, 3]
lst.clear()
print(lst)  # []
```

```python
# index(x[, start[, end]])
# Returns the index of the first occurrence of x.

lst = [1, 2, 3, 2]
print(lst.index(2))        # 1
print(lst.index(2, 2))     # 3
```

```python
# count(x)
# Returns how many times x appears.

lst = [1, 2, 2, 3]
print(lst.count(2))  # 2
```

```python
# Not recommended
m = l
m[0] = 0
print(m)
print(l)
print()
m[0]=1
print(m)
print(l)
print()
l[1] = 2
print(m)
print(l)
print()
l[1]=3
print(m)
print(l)
```

```python
# Recommended
# Returns a shallow copy of the list.

m = l.copy()
print("Original : ",l)
m[0] = 0
print(m)
print(l)
print()
l[0]=1
print(m)
print(l)
print()
```

```python
# Recursion kind of situtation 
m = [900,1000,1100]
m.append(m)
print(m)
```

```python
n = [900,1000,1100]
n.append(n[:])       # Shallow copy
print(n)
# or
import copy
o = [900,1000,1100]
o.append(copy.deepcopy(o))  # Deep copy
print(o)

```

```python
# Extend vs append
l1=[1,2,3]
l2=[4,5,6]
l1.extend(l2)
print(l1)

l1=[1,2,3]
l2=[4,5,6]
l1.append(l2)
print(l1)
```

```python
# | Method      | Mutates Original List | Returns Value |
# | ----------- | --------------------- | ------------- |
# | `append()`  | ✅ Yes                 | ❌ None        |
# | `extend()`  | ✅ Yes                 | ❌ None        |
# | `insert()`  | ✅ Yes                 | ❌ None        |
# | `remove()`  | ✅ Yes                 | ❌ None        |
# | `pop()`     | ✅ Yes                 | ✅ Element     |
# | `clear()`   | ✅ Yes                 | ❌ None        |
# | `index()`   | ❌ No                  | ✅ Index       |
# | `count()`   | ❌ No                  | ✅ Count       |
# | `sort()`    | ✅ Yes                 | ❌ None        |
# | `reverse()` | ✅ Yes                 | ❌ None        |
# | `copy()`    | ❌ No                  | ✅ New list    |
```
