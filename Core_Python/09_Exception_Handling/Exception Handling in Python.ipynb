```python
def f():
    try:
        # First error (initial trigger)
        print("Step 1: Trying to convert 'abc' to int")
        int("abc")  # This raises a ValueError
    except ValueError as ve:
        print("Caught ValueError:", ve)
        
        try:
            # Second error (happens while handling the first one)
            print("Step 2: Trying to divide 1 by 0")
            1 / 0  # This raises ZeroDivisionError
        except ZeroDivisionError as zde:
            print("Caught ZeroDivisionError:", zde)
            
            # Now raise a new RuntimeError explicitly from zde
            print("Step 3: Raising RuntimeError from ZeroDivisionError")
            raise RuntimeError("Custom runtime failure") from zde

try:
    f()
except RuntimeError as re:
    print("\n=== Final Output ===")
    print("Caught Exception:", re)
    print("Explicit Cause (from):", re.__cause__)
    print("Implicit Context (previous exception):", re.__context__)
```

```python
def f():
    try:
        open("nonexistent.txt")  # This raises FileNotFoundError
    except FileNotFoundError as file_error:
        try:
            int("not-a-number")  # This raises ValueError
        except ValueError as val_error:
            # ⛓️ Explicitly say: this RuntimeError is due to the file_error
            raise RuntimeError("Something went wrong!") from file_error

try:
    f()
except RuntimeError as e:
    print("Caught:", e)
    print("Cause (__cause__):", repr(e.__cause__))
    print("Context (__context__):", repr(e.__context__))
```
