```python
# Why os.path is useful?
import os

file_path = "e:/Python/clutteredFolder/Day1/test.txt"
print(os.path.basename(file_path))  # test.txt
print(os.path.dirname(file_path))   # e:/Python/clutteredFolder/Day1
print(os.path.splitext(file_path))  # ('e:/Python/clutteredFolder/Day1/test', '.txt')
```

```python
# Combined example 1

import os

if(not os.path.exists("clutteredFolder")):
    os.mkdir("clutteredFolder")
else:
    print("Directory exists")

for i in range(0, 3):
    os.mkdir(f"clutteredFolder/Day{i+1}")
    print(f"Created Day{i+1} in clutteredFolder")
```

```python
# Combined example 2

import os

# List the contents of the current working directory
print("Current Working Directory:", os.getcwd())
os.chdir(r"e:\Python\clutteredFolder")
folders = os.listdir(".")  # List folders/files in the current working directory

print("Contents of current directory:", folders)

for i in range(0, 100):
    old_path = f"Day{i+1}"  # Reference files directly
    new_path = f"Day {i+1}"
    
    # Check if the file exists before renaming
    if os.path.exists(old_path):
        os.rename(old_path, new_path)
        print(f"Renamed: '{old_path}' -> '{new_path}'")
    else:
        print(f"Skipping: '{old_path}' does not exist.")
```

```python
# Combined example 3

import os 
folders = os.listdir(r"e:\Python\clutteredFolder")

print(os.getcwd())
os.chdir(r"e:\Python")
print(os.getcwd())

for folder in folders:
    print(folder)
    print(os.listdir(f"clutteredFolder/{folder}"))
```
