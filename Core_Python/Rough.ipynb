```python
class Member:
    def __init__(self, member_id, name, email):
        
        self.member_id = member_id
        self.name = name
        self.email = email
        self.borrowed_books = [] 

    def borrow_book(self, ISBN):
        
        self.borrowed_books.append(ISBN)

    def return_book(self, ISBN):
        
        if ISBN in self.borrowed_books:
            self.borrowed_books.remove(ISBN)
        else:
            print(f"Book with ISBN {ISBN} is not in the member's borrowed list.")


class Library:
    def __init__(self):
        
        self.books = []  
        self.members = {} 

    def add_book(self, title, author, ISBN):
        
        book = {
            'title': title,
            'author': author,
            'ISBN': ISBN,
            'status': 'available'
        }
        self.books.append(book)
        print(f"Book '{title}' added successfully.")

    def register_member(self, member_id, name, email):
        
        if member_id in self.members:
            print(f"Member ID {member_id} already exists.")
        else:
            self.members[member_id] = Member(member_id, name, email)
            print(f"Member '{name}' registered successfully.")

    def lend_book(self, ISBN, member_id):
        
        book = next((b for b in self.books if b['ISBN'] == ISBN), None)
        if not book:
            print(f"Book with ISBN {ISBN} not found.")
            return

        if book['status'] == 'borrowed':
            print(f"Book '{book['title']}' is already borrowed.")
            return

        member = self.members.get(member_id)
        if not member:
            print(f"Member ID {member_id} not found.")
            return

        book['status'] = 'borrowed'
        member.borrow_book(ISBN)
        print(f"Book '{book['title']}' lent to member '{member.name}'.")

    def return_book(self, ISBN, member_id):
        
        book = next((b for b in self.books if b['ISBN'] == ISBN), None)
        if not book:
            print(f"Book with ISBN {ISBN} not found.")
            return

        member = self.members.get(member_id)
        if not member:
            print(f"Member ID {member_id} not found.")
            return

        if ISBN not in member.borrowed_books:
            print(f"Member '{member.name}' has not borrowed this book.")
            return

        book['status'] = 'available'
        member.return_book(ISBN)
        print(f"Book '{book['title']}' returned by member '{member.name}'.")

    def display_books(self):
      
        print("\nLibrary Books:")
        print("--------------")
        for book in self.books:
            print(f"Title: {book['title']}, Author: {book['author']}, "
                  f"ISBN: {book['ISBN']}, Status: {book['status']}")
        print("--------------\n")


if __name__ == "__main__":
    library = Library()

    library.add_book("The Great Gatsby", "F. Scott Fitzgerald", "9780743273565")
    library.add_book("To Kill a Mockingbird", "Harper Lee", "9780061120084")
    library.add_book("1984", "George Orwell", "9780451524935")

    library.register_member("M001", "Abhishek Gupta", "abhishek@example.com")
    library.register_member("M002", "Aryan somani", "aryan@example.com")

    library.display_books()

    library.lend_book("9780743273565", "M001")  
    library.lend_book("9780451524935", "M002")

    library.lend_book("9780451524935", "M001")

    library.display_books()

    library.return_book("9780743273565", "M001") 

    library.return_book("9780743273565", "M002") 

    library.display_books()
```
