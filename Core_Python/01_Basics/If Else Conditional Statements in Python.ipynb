# Basic if, elif, else

- elif = "else if"
- Python supports multiple elifs
- Indentation is mandatory

```python
x = 10

if x > 0:
    print("Positive")
elif x == 0:
    print("Zero")
else:
    print("Negative")
```

# Ternary Conditional Operator (x if condition else y)

- For short, single-line decisions
- Avoid nesting ternary expressions for clarity

```python
x = 5
status = "Even" if x % 2 == 0 else "Odd"
print(status)  # Odd
```

```python
status = lambda x:"Odd" if x%2 != 0 else "Even"
print(status(6))
```

# Conditional Expressions in Return/Assignment

Clean one-liner replacements for simple if-else functions

```python
def max(a, b):
    return a if a > b else b

val = max(10, 20)  # 20
```

# Nested if Statements

- Over-nesting reduces readability
- Prefer logical operators where possible



```python
x = 15

if x > 0:
    if x < 100:
        print("Positive and less than 100")
```

# Using if with Functions, Lambdas, Loops
Inside lambdas (limited form only):

```python
f = lambda x: "Even" if x % 2 == 0 else "Odd"
print(f(3))  # Odd
```

In loops:

```python
for i in range(5):
    if i % 2 == 0:
        print(f"{i} is even")
    else:
        print(f"{i} is odd")
```

# Using Dictionaries to Replace if-elif-else (Function Dispatching)

- Cleaner
- Easier to extend
- Functions are first-class citizens

```python
def add(): print("Add")
def sub(): print("Subtract")
def mul(): print("Multiply")

action = {
    'a': add,
    's': sub,
    'm': mul
}

choice = 's'
action.get(choice, lambda: print("Invalid"))()  # Subtract
```

# Match-Case (Python 3.10+ — Structural Pattern Matching)

- Replaces classic switch-case from other languages
- Supports complex patterns (like destructuring tuples)

```python
def switch_demo(value):
    match value:
        case 1:
            print("One")
        case 2:
            print("Two")
        case _:
            print("Default")

switch_demo(2)  # Two
```

# Conditional Expressions in List Comprehensions

```python
nums = [1, 2, 3, 4]
result = ["even" if x % 2 == 0 else "odd" for x in nums]
print(result)  # ['odd', 'even', 'odd', 'even']
```

# Boolean Shortcut (Short-Circuiting)

```python
x = None
name = x or "Default"
print(name)  # "Default" if x is falsy
```

```python
y = 10
z = y and "Yes"
print(z)  # "Yes" because y is truthy
```

```python
def describe(value):
    match value:
        case int():
            print("It's an integer")
        case str():
            print("It's a string")
        case list():
            print("It's a list")
        case _:
            print("Unknown type")

describe(10)        # It's an integer
describe("hello")   # It's a string
```

```python
data = [1, 2]

match data:
    case [1, 2]:
        print("Exact match [1, 2]")
    case [1, *rest]:
        print("Starts with 1:", rest)
    case [a, b]:
        print("Two items:", a, b)
    case _:
        print("No match")
```

```python
data = {"type": "error", "code": 404}

match data:
    case {"type": "error", "code": code}:
        print(f"Error code: {code}")
    case {"type": "success", "data": _}:
        print("Success response")
    case _:
        print("Unknown format")
```

```python
x = "yes"

match x:
    case "yes" | "y" | "yeah":
        print("Affirmative")
    case "no" | "n":
        print("Negative")
    case _:
        print("Unknown")
```

```python
class Point:
    def __init__(self, x, y): self.x, self.y = x, y

# p = Point(1, 2)
p = Point(0, 0)

match p:
    case Point(x=0, y=0):
        print("Origin")
    case Point(x=x, y=y):
        print(f"Point at ({x}, {y})")
    case _:
        print("Unknown object")
```

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

p = Point(0, 0)

match p:
    case Point(0, 0):
        print("Origin")
    case Point(x, y):
        print(f"Point({x}, {y})")
```
